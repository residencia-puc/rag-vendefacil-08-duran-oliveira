## Structured

### `customers.csv` | Cadastro de clientes da VendeFácil

Tem 2000 registros, 11 colunas, sem valores nulos.

Descrição geral: 

* dados cadastrais e comerciais dos clientes (empresas) que usam produto VendeFácil. 

* Fonte tabular, cada linha é um registro único de cliente. 

* Uso típico no RAG: perguntas por estado, segmento, plano, produto contratado ou status da conta.

<div style="font-size: 13px">

| Campo | Tipo | Descrição | Valores / exemplos |
|---|---|---|---|
| `customer_id` | string | Identificador único do cliente | `CUST001`, `CUST002`, ... |
| `company_name` | string | Razão social / nome fantasia da empresa cliente | `Supermercado Boa Compra` |
| `cnpj` | string | CNPJ da empresa (dado sensível) | `12.345.678/0001-90` |
| `state` | string | UF onde o cliente está sediado | BA, CE, DF, ES, GO, MG, PE, PR, RJ, RS, SC, SP (12 estados) |
| `city` | string | Cidade do cliente | Belo Horizonte, Campinas, ... |
| `segment` | string | Segmento/ramo de atuação do cliente | Supermercado, Farmácia, Alimentação, Vestuário, Automotivo, Calçados, Construção, Conveniência, Cosméticos, Eletrônicos, Pet, Saúde/Ótica, Varejo Cultural (13 valores) |
| `plan` | string | Plano contratado | Basic, Pro, Enterprise |
| `main_product` | string | Produto VendeFácil principal usado | VendeFácil PDV, VendeFácil Loja, VendeFácil Estoque, VendeFácil Analytics, VendeFácil Pay |
| `mrr` | float | Receita recorrente mensal (R$) | 300.01 a 3498.85 |
| `status` | string | Situação da conta | Ativo, Cancelado, Inativo |
| `contact_email` | string | E-mail de contato comercial (dado pessoal) | `gerencia@boacompra.com.br` |

</div>

### products.json

Descrição geral: 

* catálogo de produtos VendeFácil (JSON estruturado). 
* Contém dois blocos: 
    * pricing_plans (regras dos 3 planos comerciais) 
    * products (5 produtos, cada um com preço, features, responsáveis técnicos e SLA). 
* Dado não sensível — uso normal, com citação.

<div style="font-size: 13px">

| Campo | Tipo | Descrição | Valores / exemplos |
|---|---|---|---|
| `company` | string | Nome da empresa (metadado do arquivo) | `VendeFácil Tecnologia Ltda.` |
| `last_updated` | string (data) | Data da última atualização do catálogo | `2026-02-01` |
| `pricing_plans.<Plano>.monthly_fee_brl` | float | Mensalidade do plano (R$) | 350.0, 890.0, 2450.0 |
| `pricing_plans.<Plano>.included_terminals` | int | Nº de terminais incluídos no plano | 1, 3, 8 |
| `pricing_plans.<Plano>.extra_terminal_fee_brl` | float | Custo do terminal adicional (R$) | 50.0, 45.0, 35.0 |
| `pricing_plans.<Plano>.description` | string | Descrição do público-alvo do plano | "Plano ideal para pequenos comércios..." |
| `product_id` | string | Identificador único do produto | `PROD-PDV`, `PROD-ESTOQUE`, ... |
| `name` | string | Nome comercial do produto | `VendeFácil PDV` |
| `category` | string | Categoria/área do produto | "Ponto de Venda & Frente de Caixa" |
| `description` | string | Descrição funcional do produto | texto livre |
| `tech_lead` | string | Responsável técnico (liga a `employees.csv`) | `Beatriz Lima` |
| `product_manager` | string | Gerente de produto (liga a `employees.csv`) | `Ana Souza` |
| `standalone_monthly_price_brl` | float | Preço avulso do produto (R$), quando aplicável | 350.0, 490.0, ... |
| `pricing_by_plan` | object | Preço/condição do produto dentro de cada plano | "Incluso no plano", "Adicional de R$..." |
| `transaction_rates` | object | Taxas por transação (apenas VendeFácil Pay) | "0.99% por transação" |
| `features` | list[string] | Lista de funcionalidades do produto | "Emissão NFC-e e SAT", ... |
| `supported_os` | list[string] | Plataformas/SOs suportados | Windows, Linux, Android, Web Cloud, iOS |
| `sla_uptime` | string | SLA de disponibilidade contratual | "99.9%" |

</div>

### Sales.csv


Descrição geral: 
* registro transacional de vendas (3000 linhas) ligando cliente, loja e produto vendido. 

* Fonte tabular — cada linha é uma transação de venda. 

<div style="font-size: 13px">

| Campo | Tipo | Descrição | Valores / exemplos |
|---|---|---|---|

| `company_name` | string | Nome da empresa cliente (redundante com `customers.csv`) | `Empório da Cunha Eireli` |
| `store_id` | string | Identificador da loja/filial (liga a `stores.json`) | `STR-005` |
| `store_name` | string | Nome da loja/filial | `Filial Shopping Rio de Janeiro` |
| `state` | string | UF da loja | 12 estados (mesmos de `customers.csv`) |
| `city` | string | Cidade da loja | `Rio de Janeiro`, `Cascavel`, ... |
| `product_id` | string | Produto vendido (liga a `products.json`) | PROD-PDV, PROD-ESTOQUE, PROD-LOJA, PROD-ANALYTICS, PROD-PAY |
| `product_name` | string | Nome do produto vendido | `VendeFácil Loja` |
| `date` | string (datetime) | Data/hora da transação | `2025-05-04 14:47:10` (jan/2025 a fev/2026) |
| `payment_method` | string | Forma de pagamento | PIX, Boleto Bancário, Cartão de Crédito, Cartão de Débito, Dinheiro |
| `amount_brl` | float | Valor da transação (R$) | 121.40 a 3099.28 |
| `pos_terminal` | string | Terminal/caixa onde ocorreu a venda | `Caixa 03` |
| `status` | string | Situação da venda | Aprovada, Cancelada, Estornada |

</div>

### stores.json

Descrição geral: 

* cadastro das 50 lojas/filiais da rede de clientes, ligando cada loja ao seu cliente (customer_id) e aos módulos VendeFácil ativos nela. 

* JSON estruturado com lista network_stores.

<div style="font-size: 13px">

| Campo | Tipo | Descrição | Valores / exemplos |
|---|---|---|---|
| `store_id` | string | Identificador único da loja/filial | `STR-001` |
| `customer_id` | string | Cliente dono da loja (liga a `customers.csv`) | `CUST001` |
| `company_name` | string | Nome da empresa cliente (redundante com `customers.csv`) | `Supermercado Boa Compra` |
| `store_name` | string | Nome da filial | `Filial Matriz - Centro BH` |
| `state` | string | UF da loja | 12 estados (mesmos das demais fontes) |
| `city` | string | Cidade da loja | `Belo Horizonte` |
| `pos_terminals_count` | int | Nº de terminais/caixas ativos na loja | 1 a 12 |
| `active_modules` | list[string] | Módulos VendeFácil ativos nessa loja | VendeFácil PDV, Estoque, Loja, Analytics, Pay |

</div>
