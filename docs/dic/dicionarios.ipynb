## Structured

### `customers.csv` | Cadastro de clientes da VendeFácil

Tem 2000 registros, 11 colunas, sem valores nulos.

Descrição geral: 

* dados cadastrais e comerciais dos clientes (empresas) que usam produto VendeFácil. 

* Fonte tabular, cada linha é um registro único de cliente. 

* Uso típico no RAG: perguntas por estado, segmento, plano, produto contratado ou status da conta.

<div style="font-size: 13px">

| Campo | Tipo | Descrição | Valores / exemplos |
|---|---|---|---|
| `customer_id` | string | Identificador único do cliente | `CUST001`, `CUST002`, ... |
| `company_name` | string | Razão social / nome fantasia da empresa cliente | `Supermercado Boa Compra` |
| `cnpj` | string | CNPJ da empresa (dado sensível) | `12.345.678/0001-90` |
| `state` | string | UF onde o cliente está sediado | BA, CE, DF, ES, GO, MG, PE, PR, RJ, RS, SC, SP (12 estados) |
| `city` | string | Cidade do cliente | Belo Horizonte, Campinas, ... |
| `segment` | string | Segmento/ramo de atuação do cliente | Supermercado, Farmácia, Alimentação, Vestuário, Automotivo, Calçados, Construção, Conveniência, Cosméticos, Eletrônicos, Pet, Saúde/Ótica, Varejo Cultural (13 valores) |
| `plan` | string | Plano contratado | Basic, Pro, Enterprise |
| `main_product` | string | Produto VendeFácil principal usado | VendeFácil PDV, VendeFácil Loja, VendeFácil Estoque, VendeFácil Analytics, VendeFácil Pay |
| `mrr` | float | Receita recorrente mensal (R$) | 300.01 a 3498.85 |
| `status` | string | Situação da conta | Ativo, Cancelado, Inativo |
| `contact_email` | string | E-mail de contato comercial (dado pessoal) | `gerencia@boacompra.com.br` |

</div>
